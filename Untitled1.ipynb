```python
players = [
    {"name": "Serena Williams", "sport": "Tennis", "achievements": 23},
    {"name": "Lionel Messi", "sport": "Soccer", "achievements": 7},
    {"name": "Michael Phelps", "sport": "Swimming", "achievements": 23},
    {"name": "Usain Bolt", "sport": "Athletics", "achievements": 8},
    {"name": "Roger Federer", "sport": "Tennis", "achievements": 20},
    {"name": "Cristiano Ronaldo", "sport": "Soccer", "achievements": 5}
]

for player in players:
    if player["name"] == "Lionel Messi":
        if player["achievements"] > 10:
            print("Name:", player["name"])
            print("Sport:", player["sport"])
            print("Achievements:", player["achievements"])
        else:
            print("Lionel Messi does not have more than 10 achievements.")
```

```python
## 2. Use of OR operator

for player in players:
    if player["sport"] == "Tennis" or player["achievements"] == 20:
        print(player["name"], "meets the condition successfully.")
```

```python
for player in players:
    if player["achievements"] < 10 and player["sport"] != "Soccer":
        print("Name:", player["name"])
        print("Sport:", player["sport"])
        print("Achievements:", player["achievements"])
        print()
```

```python
num = int(input("Enter an integer between 1 and 9999: "))

# Check range
if num >= 1 and num <= 9999:
    # 1. Even or Odd
    if num % 2 == 0:
        print("The number is Even.")
    else:
        print("The number is Odd.")

    # 2. Palindrome
    if str(num) == str(num)[::-1]:
        print("The number is a Palindrome.")
    else:
        print("The number is not a Palindrome.")

    # 3. Divisible by sum of digits
    digit_sum = sum(int(digit) for digit in str(num))
    if num % digit_sum == 0:
        print("The number is divisible by the sum of its digits.")
    else:
        print("The number is not divisible by the sum of its digits.")
else:
    print("Please enter a number between 1 and 9999.")
```

```python
squares = ['red', 'yellow', 'green', 'purple', 'blue']
new_squares = []
i = 0

while i < len(squares):
    if squares[i] == 'orange':
        new_squares.append(squares[i])
    else:
        break
    i += 1

print("New list:", new_squares)
```

```python
Animals = ['lion', 'giraffe', 'gorilla', 'parrots', 'crocodile', 'deer', 'swan']
seven_letter_animals = []
i = 0

while i < len(Animals):
    if len(Animals[i]) == 7:
        seven_letter_animals.append(Animals[i])
    i += 1

print("Animals with 7 letters:")
print(seven_letter_animals)
```
