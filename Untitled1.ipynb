```python
#create the dataset in a separate cell.
players = [
    ["Serena Williams", "Tennis", 23, "Grand Slams"],
    ["Lionel Messi", "Soccer", 7, "Ballon d'Ors"],
    ["Michael Phelps", "Swimming", 23, "Gold Medals"],
    ["Usain Bolt", "Athletics", 8, "Gold Medals"],
    ["Roger Federer", "Tennis", 20, "Grand Slams"],
    ["Cristiano Ronaldo", "Soccer", 5, "Ballon d'Ors"]
]

print("Player Name\t\tSport\t\tAchievements")

for player in players:
    print(player[0], "\t", player[1], "\t", player[2], player[3])
```

```python
#Check whether Lionel Messi has more than 10 achievements.
for player in players:
    if player[0] == "Lionel Messi":
        if player[2] > 10:
            print("Player Name:", player[0])
            print("Sport:", player[1])
            print("Achievements:", player[2], player[3])
        else:
            print("Does not have more than 10 achievements.")
```

```python
#Check whether a player belongs to Tennis or has exactly 20 achievements.
for player in players:
    if player[1] == "Tennis" or player[2] == 20:
        print(player[0], "meets the condition.")
```

```python
#Check whether a player has less than 10 achievements and does not play Soccer.
for player in players:
    if player[2] < 10 and player[1] != "Soccer":
        print("Player Name:", player[0])
        print("Sport:", player[1])
        print("Achievements:", player[2], player[3])
```

```python
#Check Even/Odd, Palindrome and Divisibility by Sum of Digits.
num = int(input("Enter a number (1 to 9999): "))

if 1 <= num <= 9999:

    # Even or Odd
    if num % 2 == 0:
        print("Even Number")
    else:
        print("Odd Number")

    # Palindrome Check
    if str(num) == str(num)[::-1]:
        print("Palindrome Number")
    else:
        print("Not a Palindrome Number")

    # Sum of Digits
    digit_sum = sum(int(digit) for digit in str(num))

    # Divisibility Check
    if num % digit_sum == 0:
        print("Divisible by the sum of its digits")
    else:
        print("Not divisible by the sum of its digits")

else:
    print("Please enter a number between 1 and 9999.")
```

```python
#Copy 'orange' strings from the list squares to new_squares using a while loop. Stop if the value is not 'orange'.
squares = ['red', 'yellow', 'green', 'purple', 'blue']

new_squares = []
i = 0

while i < len(squares):
    if squares[i] != 'orange':
        break

    new_squares.append(squares[i])
    i += 1

print("New Squares:", new_squares)
```

```python
#Find animals whose names contain exactly 7 letters using a while loop.
Animals = ["lion", "giraffe", "gorilla", "parrots",
           "crocodile", "deer", "swan"]

new_animals = []
i = 0

while i < len(Animals):
    if len(Animals[i]) == 7:
        new_animals.append(Animals[i])

    i += 1

print("Animals with 7 letters:", new_animals)
```
