```python
#Problem 1: Create a DataFrame
import pandas as pd
import numpy as np

data = {
    'Student': ['Tanvi', 'Mahima', 'Prince', 'Satyam'],
    'Age': [18, 42, 20, 21],
    'Country': ['India', 'India', 'India', 'India'],
    'Course': ['CSE AI', 'Fashion', 'Python', 'Data Science'],
    'Marks': [95, 88, 92, 85]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Problem 2: Retrieve the Marks Column
b = df['Marks']
print(b)
```

```python
#Problem 3: Retrieve Country and Course Columns
c = df[['Country', 'Course']]
print(c)
```

```python
#Problem 4: Data Loading and Exploration
data = {
    'Customer_ID': range(1, 11),
    'Name': ['Tanvi', 'Mahima', 'Prince', 'Satyam', 'Himanshu',
             'Mridul', 'Riya', 'Ananya', 'Rahul', 'Simran'],
    'Age': [18, 42, 20, 21, 22, 20, 19, 23, 25, 24],
    'Gender': ['Female', 'Female', 'Male', 'Male', 'Male',
               'Male', 'Female', 'Female', 'Male', 'Female'],
    'Purchase_Amount': [2000, 3500, 1500, 4000, 5000,
                        2500, 3000, 1800, 4500, 2200]
}

df = pd.DataFrame(data)

print("First 5 Rows:")
print(df.head())

print("\nSummary Statistics:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

df = df.fillna(0)
print("\nData After Handling Missing Values:")
print(df)
```

```python
#Problem 5: Data Filtering and Selection

# Customers whose age is greater than 30
print(df[df['Age'] > 30])

# Total purchase amount by gender
print(df.groupby('Gender')['Purchase_Amount'].sum())

# Select specific columns
print(df[['Customer_ID', 'Name', 'Purchase_Amount']])
```

```python
#Problem 6: Data Transformation and Aggregation
# Apply a 10% discount
df['Discounted_Price'] = df['Purchase_Amount'] * 0.90

print("Data with Discounted Price:")
print(df)

# Average purchase amount by gender
print("\nAverage Purchase Amount by Gender:")
print(df.groupby('Gender')['Purchase_Amount'].mean())

# Sort in descending order
print("\nSorted Dataset:")
print(df.sort_values(
    by='Purchase_Amount',
    ascending=False
))
```
