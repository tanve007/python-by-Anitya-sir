```python
# Float to Integer
float_value = 10.8
integer_from_float = int(float_value)

# String to Integer
string_value = "25"
integer_from_string = int(string_value)

# Boolean to Integer
boolean_value = True
integer_from_boolean = int(boolean_value)

print("Float to Integer:", integer_from_float)
print("String to Integer:", integer_from_string)
print("Boolean to Integer:", integer_from_boolean)
```
