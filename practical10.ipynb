```python
#1. Import Libraries and Load Dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df['species'] = pd.Categorical.from_codes(
    iris.target, iris.target_names
)

# Display dataset
print(df.head())

# Dataset information
print("\nDataset Information:")
print(df.info())

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

```python
#2. Scatter Plots
#Aim: To show the relationship between two numerical variables using scatter plots.
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.scatterplot(
    data=df,
    x='sepal length (cm)',
    y='sepal width (cm)',
    hue='species'
)
plt.title("Sepal Length vs Sepal Width")

plt.subplot(1, 2, 2)
sns.scatterplot(
    data=df,
    x='petal length (cm)',
    y='petal width (cm)',
    hue='species'
)
plt.title("Petal Length vs Petal Width")

plt.tight_layout()
plt.show()
```

Analysis:
- Setosa has smaller petal dimensions than the other species.
- Versicolor and Virginica show larger petal measurements.
- Petal length and petal width have a positive relationship.
- Different species form separate clusters.

```python
# 3. Density Plots
# Aim: To visualize the distribution of data using density curves.

features = iris.feature_names

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for feature, ax in zip(features, axes.flatten()):
    sns.kdeplot(
        data=df,
        x=feature,
        hue='species',
        fill=False,
        ax=ax
    )
    ax.set_title(f"Density Plot of {feature}")

plt.tight_layout()
plt.show()
```

Analysis:
- Density plots show the distribution of each flower measurement.
- Setosa has a different distribution of petal length and width compared to other species.
- Versicolor and Virginica have some overlapping distributions.
- These plots help us understand the variation in flower measurements.

```python
# 4. Violin Plots
# Aim: To compare the distribution and spread of numerical data across different species.
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for feature, ax in zip(features, axes.flatten()):
    sns.violinplot(
        data=df,
        x='species',
        y=feature,
        ax=ax
    )
    ax.set_title(f"Violin Plot of {feature}")

plt.tight_layout()
plt.show()
```

Analysis:
- Violin plots show the spread and distribution of measurements.
- Setosa has smaller petal length and petal width.
- Virginica generally has larger petal measurements.
- The width of each violin represents the density of observations.

```python
# 5. Heatmap
# Aim: To identify the correlation between numerical variables using a heatmap.
# Calculate correlation
correlation = df.drop(columns='species').corr()

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Correlation Heatmap of Iris Dataset")
plt.show()
```

Analysis (Correlation):
- Petal length and petal width have a very strong positive correlation (approximately 0.96).
- Sepal length and petal length also have a strong positive correlation (approximately 0.87).
- Sepal width and petal width have a negative correlation (approximately -0.37).
- A positive correlation means both variables tend to increase together.
- A negative correlation means one variable tends to decrease when the other increases.

```python
# 6. Joint Plots
# Aim: To visualize the relationship between two variables along with their individual distributions.

sns.jointplot(
    data=df,
    x='sepal length (cm)',
    y='sepal width (cm)',
    hue='species',
    kind='scatter',
    height=7
)

plt.show()
```

Analysis:
- Joint plots show the relationship between sepal length and sepal width.
- The central scatter plot displays the relationship between both variables.
- The top and right sides show their individual distributions.
- Different species show different patterns, with some overlapping observations.

```python
# 7. Pair Plots
# Aim: To visualize pairwise relationships between all numerical variables in the dataset.
sns.pairplot(
    df,
    hue='species',
    diag_kind='kde'
)

plt.show()
```

Analysis:
- Pair plots show relationships between all numerical features.
- Setosa is clearly separated from other species based on petal measurements.
- Versicolor and Virginica show some overlap.
- Petal length and petal width are strongly related.
- Pair plots help compare multiple variables at the same time.

Final Conclusion:

In this practical, we used the Seaborn library to visualize the Iris dataset using scatter plots, density plots, violin plots, heatmaps, joint plots and pair plots. These visualizations helped us understand data distribution, relationships and correlations between different flower measurements. We observed that petal length and petal width have a strong positive correlation and can help distinguish different Iris species.
