```python
shopping_list = []

print("Empty shopping list:", shopping_list)
```

```python
shopping_list = ["Watch", "Laptop", "Shoes", "Pen", "Clothes"]

print("Shopping list:", shopping_list)
print("Number of items:", len(shopping_list))
```

```python
shopping_list.append("Football")

print("Shopping list after adding Football:", shopping_list)
```

```python
print("Last item:", shopping_list[-1])
```

```python
print("Entire shopping list:")
print(shopping_list)
```

```python
print("Important items:")
print(shopping_list[1])
print(shopping_list[2])
```

```python
shopping_list[3] = "Notebook"

print("Shopping list after replacing Pen with Notebook:")
print(shopping_list)
```

```python
shopping_list.remove("Watch")

print("Final shopping list:")
print(shopping_list)
```
