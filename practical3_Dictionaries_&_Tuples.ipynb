```python
#Create an empty Dictionary
inventory = {}

print("Empty Inventory:", inventory)
```

```python
#Store the first product details in variables
ProductNo1_Name = "Mobile Phone"
ProductNo1_Quantity = 5
ProductNo1_Price = 20000
ProductNo1_releaseYear = 2020

print("Product 1 Name:", ProductNo1_Name)
print("Product 1 Quantity:", ProductNo1_Quantity)
print("Product 1 Price:", ProductNo1_Price)
print("Product 1 Release Year:", ProductNo1_releaseYear)
```

```python
#Add first product details to inventory
inventory["ProductNo1"] = {
    "Product Name": ProductNo1_Name,
    "Quantity": ProductNo1_Quantity,
    "Price": ProductNo1_Price,
    "Release Year": ProductNo1_releaseYear
}

print("Inventory:", inventory)
```

```python
#Store the second product details in variables
ProductNo2_Name = "Laptop"
ProductNo2_Quantity = 3
ProductNo2_Price = 50000
ProductNo2_releaseYear = 2022

print("Product 2 Name:", ProductNo2_Name)
print("Product 2 Quantity:", ProductNo2_Quantity)
print("Product 2 Price:", ProductNo2_Price)
print("Product 2 Release Year:", ProductNo2_releaseYear)
```

```python
#Add second product details to inventory
inventory["ProductNo2"] = {
    "Product Name": ProductNo2_Name,
    "Quantity": ProductNo2_Quantity,
    "Price": ProductNo2_Price,
    "Release Year": ProductNo2_releaseYear
}

print("Inventory after adding Product 2:")
print(inventory)
```

```python
#Display the products present in inventory
print("Products present in inventory:")

for product_no, details in inventory.items():
    print(product_no, ":", details["Product Name"])
```

```python
#Check if release years are in inventory
print("Checking release years:")

print("ProductNo1 release year is in inventory:",
      ProductNo1_releaseYear == inventory["ProductNo1"]["Release Year"])

print("ProductNo2 release year is in inventory:",
      ProductNo2_releaseYear == inventory["ProductNo2"]["Release Year"])
```

```python
#Delete release year of both products
del inventory["ProductNo1"]["Release Year"]
del inventory["ProductNo2"]["Release Year"]

print("Inventory after deleting release years:")
print(inventory)
```

```python
#Create the tuple
prices = (250, 300, 150, 400, 100, 350, 200)

print("Prices:", prices)
```

```python
#Find highest and lowest price
highest_price = max(prices)
lowest_price = min(prices)

print("Highest Price:", highest_price)
print("Lowest Price:", lowest_price)
```

```python
#Calculate total sum
total = sum(prices)

print("Total sum of all prices:", total)
```

```python
#Convert tuple into sorted list
sorted_prices = sorted(prices)

print("Sorted list:", sorted_prices)
```

```python
#try to modify an element
try:
    prices[0] = 500
except TypeError as e:
    print("Error:", e)
    print("Explanation: Tuples are immutable, so their elements cannot be changed.")
```
