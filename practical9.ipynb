```python
#Given a filename and a list of strings, create a file containing zippered pairs of strings, with each pair written on a separate line.
def zipper_file(filename, strings):
    with open(filename, "w") as file:
        for i in range(0, len(strings), 2):
            str1 = strings[i]
            str2 = strings[i + 1]

            zippered = ""

            for a, b in zip(str1, str2):
                zippered += a + b

            zippered += str1[len(str2):] if len(str1) > len(str2) else str2[len(str1):]

            file.write(zippered + "\n")


filename = "zipper.txt"

strings = ["abc", "123", "hello", "world"]

zipper_file(filename, strings)

print("File created successfully!")

with open(filename, "r") as file:
    print(file.read())
```

```python
#Given a list of dates in MM/DD/YYYY format, convert them into English sentences in the format Office Hours on Month DD, YYYY.
def reformat_date(dates):

    months = {
        "01": "January",
        "02": "February",
        "03": "March",
        "04": "April",
        "05": "May",
        "06": "June",
        "07": "July",
        "08": "August",
        "09": "September",
        "10": "October",
        "11": "November",
        "12": "December"
    }

    result = []

    for date in dates:
        month, day, year = date.split("/")

        formatted_date = "Office Hours on " + months[month] + " " + day + ", " + year

        result.append(formatted_date)

    return result


dates = ["01/15/2025", "06/09/2025", "12/25/2025"]

output = reformat_date(dates)

for date in output:
    print(date)
```
